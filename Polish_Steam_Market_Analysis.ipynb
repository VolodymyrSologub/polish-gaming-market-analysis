```python
import pandas as pd

# Вказуємо шлях до файлу (шлях залежить від того, в якій папці лежить файл)
path = '/content/drive/MyDrive/games.csv'

# Читаємо файл
df = pd.read_csv(path)

# Виводимо інформацію про колонки та перші рядки
print(df.info())
df.head()
```

```python
## A Look at the Polish Game Development Industry
```

```python
# List of well-known Polish studios
polish_studios = [
    'CD PROJEKT RED', 'Techland', '11 bit studios', 'Flying Wild Hog',
    'People Can Fly', 'Bloober Team', 'The Farm 51', 'CI Games',
    'Creepy Jar', 'Frozen District', 'Starward Industries', 'Teyon',
    'PlayWay S.A.', 'Rebel Wolves', 'Fool\'s Theory', '11 bit'
]

# Filtering the dataset
# We capitalize the names so that the search is accurate regardless of capitalization
df_poland = df[df['Developers'].str.contains('|'.join(polish_studios), case=False, na=False)].copy()

print(f"Polish games found: {len(df_poland)}")

# Let's list the top 10 Polish games by number of positive reviews
if len(df_poland) > 0:
    display(df_poland[['Name', 'Developers', 'Price', 'Positive', 'Negative']].sort_values(by='Positive', ascending=False).head(10))
else:
    print("No games found. Let's try another list of studios?")
```

```python
## Revenue Estimation
```

```python
# Review rate (on average, 1 review per 30 sales)
reviews_multiplier = 30

# 1. Counting the number of owners (Estimated Owners)
df_poland['Estimated_Sales'] = (df_poland['Positive'] + df_poland['Negative']) * reviews_multiplier

# 2. Calculating approximate gross revenue
# Note: The price in the dataset may be in cents or dollars;  check the output.
# If the price is 90 instead of 0.90, divide by 100 to get the price in USD.
df_poland['Estimated_Revenue_USD'] = df_poland['Estimated_Sales'] * (df_poland['Price'] / 100)

# 3. Let's take a look at the top 10 most profitable games
top_revenue = df_poland[['Name', 'Developers', 'Price', 'Estimated_Sales', 'Estimated_Revenue_USD']].sort_values(by='Estimated_Revenue_USD', ascending=False).head(10)

# We format the output so that the numbers are readable (with commas)
pd.options.display.float_format = '{:,.2f}'.format
display(top_revenue)
```

```python
## An Analysis of the “Kings” of the Polish Game Development Industry
```

```python
# Group the data by developer
studio_analysis = df_poland.groupby('Developers').agg({
    'Name': 'count',
    'Estimated_Sales': 'sum',
    'Estimated_Revenue_USD': 'sum'
}).rename(columns={'Name': 'Games_Count'}).sort_values(by='Estimated_Revenue_USD', ascending=False)

# Display the result
print("Revenue Analysis by Studio (Estimated):")
display(studio_analysis)
```

```python
import matplotlib.pyplot as plt

# Style Settings
plt.style.use('ggplot')
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 8))

# --- CHART 1: Pie Chart (Market Share) ---
# We'll take the top 5 studios and group the rest under “Others”
top5 = studio_analysis.head(5)
others_revenue = studio_analysis['Estimated_Revenue_USD'].iloc[5:].sum()
pie_data = pd.concat([top5['Estimated_Revenue_USD'], pd.Series({'Others': others_revenue})])

ax1.pie(pie_data, labels=pie_data.index, autopct='%1.1f%%', startangle=140, colors=plt.cm.Paired.colors)
ax1.set_title('Market Share by Revenue (Polish Studios)', fontsize=16, fontweight='bold')

# --- CHART 2: Scatter Plot (Price vs. Positive Reviews) ---
# We'll focus on games priced under $100 to keep things manageable
df_filtered = df_poland[df_poland['Price'] < 10000] # if the price is in cents

ax2.scatter(df_filtered['Price']/100, df_filtered['Positive'], alpha=0.5, c='green', edgecolors='black')
ax2.set_title('Price vs. Positive Reviews', fontsize=16, fontweight='bold')
ax2.set_xlabel('Price (USD)', fontsize=12)
ax2.set_ylabel('Positive Reviews Count', fontsize=12)
ax2.set_yscale('log') # We're using a logarithmic scale because The Witcher/Cyberpunk are too high

plt.tight_layout()
plt.show()
```

```python
# Select only the columns that are working properly (Studio, Price, Reviews, Revenue)
# We are ignoring the “Name” column because it is incorrect, and we will use the studio names instead
export_data = df_poland[['Developers', 'Price', 'Positive', 'Negative', 'Estimated_Sales', 'Estimated_Revenue_USD']].copy()

# Let's save this to a new file on Google Drive
export_path = '/content/drive/MyDrive/Polish_Games_Market_Clean.csv'
export_data.to_csv(export_path, index=False)

print(f"All done! The clean file has been saved here: {export_path}")
```
